```python
import os
import pandas as pd

# Set the directory containing the CSV files
directory = "path/to/your/csv/files"  # Change to your actual directory path

directory = directory = 'DocLayoutBase1.csv/csv/1'
# Dictionary to store dataframes
dataframes = {}

# Loop through each file in the directory
for filename in os.listdir(directory):
    if filename.endswith(".csv"):  # Check if file is a CSV
        # Split the filename by hyphen
        parts = filename.split("-")
        
        # Ensure there are enough parts in the filename
        if len(parts) > 1:
            name = parts[1]  # Extract the text between the first and second hyphen
            
            # Read the CSV into a dataframe, ensuring no column is treated as an index
            file_path = os.path.join(directory, filename)
            df = pd.read_csv(file_path, index_col=0)  # Read CSV without setting an index
            
            # Store in the dictionary with extracted name
            dataframes[name] = df

# Create a new DataFrame to store combined data
combined_df = pd.DataFrame()

# Loop through each dataframe and extract the second column
for name, df in dataframes.items():
    if df.shape[1] > 1:  # Ensure there are at least two columns
        second_col_name = df.columns[2]  # Get the name of the second column
        temp_df = pd.DataFrame({name: df[second_col_name]})  # Create a temporary DataFrame
        combined_df = pd.concat([combined_df, temp_df], axis=1)  # Merge along columns

# Display the final combined DataFrame
print(combined_df.head())  # Shows the first few rows
```
